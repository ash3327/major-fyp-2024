```python
import json

# Calculate mean embeddings for each class
class_means = {}
for label in np.unique(labels):
    class_features = features[labels == label]
    class_means[label] = np.mean(class_features, axis=0).tolist()  # Convert to list for JSON

# Save means to file
with open('class_means.json', 'w') as f:
    json.dump(class_means, f)
```

```python
{k: str(v) for k, v in class_means.items()}
```

```python
def find_closest_mean(embedding, class_means, threshold=None):
    """Find the closest class mean to the given embedding"""
    min_dist = float('inf')
    closest_class = None
    for label, mean in class_means.items():
        mean_array = np.array(mean)  # Convert back from list
        dist = np.linalg.norm(embedding - mean_array)
        if dist < min_dist:
            min_dist = dist
            closest_class = label
    if threshold is not None and min_dist > threshold:
        closest_class = "No Class"
    return closest_class, min_dist
```

```python
# Calculate predictions using closest mean
predictions = []
min_dists = []
for feature in features:
    pred, min_dist = find_closest_mean(feature, class_means)
    predictions.append(pred)
    min_dists.append(min_dist)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(labels, predictions)
print(f"Classification Accuracy: {accuracy:.4f}")
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Generate confusion matrix
cm = confusion_matrix(labels, predictions)

# Plot confusion matrix
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()
```
